```python
import cv2
import numpy as np
from glob import glob
```

```python
cams = glob("/dev/video-i*")
cam_dev = cams[0]
print(f"using cam device {cam_dev}") 
res = (1280, 720)
fps=30
cam = cv2.VideoCapture(f'v4l2src device={cam_dev} ! video/x-raw,width={res[0]},height={res[1]},framerate={fps}/1 ! imxvideoconvert_g2d ! video/x-raw,format=(string)RGBA ! appsink', cv2.CAP_GSTREAMER)
```

```python
from PIL import Image
from IPython.display import display

ret, img = cam.read()
if img is not None:
    display(Image.fromarray(img))
else: 
    print("something went wrong")
```
